```javascript
// create permutations
function createConsoCombi(combinations, consonants) {
    var conso_combi = [];
    \$.each(combinations, function (_1, combi) {
        $.each(consonants, function (_2, conso) {
            var cc = [conso[0] + combi[2], combi[0] + conso[1] + combi[1]];
            if (conso.length > 2) { // add one-way direction if any
                cc.push(conso[2]);
            }
            conso_combi.push(cc);
        });
    });
    return conso_combi;
}
```
ก็คือการ loop ทั้ง combination และ consonant แล้วสร้าง permu จากการเอา combi(สระ) มาผสมใส่ consonant
เช่น (k e, เ ก .) คือใส่สระได้ทั้งหน้าและหลัง และถ้ามี conso[2] ก็ใส่ด้วย?


```javascript
function genericConvert(dir, conso_combi, specials) {
    conso_combi.sort(function(a, b) {
        return b[dir].length - a[dir].length;  // พยายาม match ที่ยาวๆ ก่อน
    });
    $.each(conso_combi, function(_1, cc) {
        if (cc.length < 3 || cc[2] == dir) {  // กรณีปกติ จะ len==2 ถ้าเป็น 3 จะบอก direction ไว้ ต้องตรงถึงจะทำงาน
            replaceRe(cc[dir], cc[+!dir]);  // ใช้ dir บอกทิศ ว่าแทนตัวแรกด้วยหลัง หรือกลับกัน
        }
    });

    specials.sort(function(a, b) {
        return b[dir].length - a[dir].length;  // เอาที่ยาวๆ ก่อนเหมือนกัน
    });
    $.each(specials, function(_1, v) {
        if (v.length < 3 || v[2] == dir) {    // สรุปคือแทนที่ธรรมดา ยกเว้นถ้ามี dir จะต้องตรง คือไปได้ทิศเดียว
            replaceRe(v[dir], v[+!dir]);
        }
    });
}
```

## DPPN
แทนทีละตัว แล้วใช้ rule-based มาแก้

````
c2t = {
    'k' : 'ก',
    'kh': 'ข',
    'g' : 'ค',
    'gh': 'ฆ',
    #''
    'c' : 'จ',
    'ch': 'ฉ',
    'j' : 'ช',
    'jh': 'ฌ',
    'ñ' : 'ญ',
    # ใน dppn รวม t, .t เข้าด้วยกัน
    't' : 'ต',
    'th': 'ถ',
    'd' : 'ท',
    'dh': 'ธ',
    'n' : 'น',
    #
    'p' : 'ป',
    'ph': 'ผ',
    'b' : 'พ',
    'bh': 'ภ',
    'm' : 'ม',
    #
    'r' : 'ร',
    'l' : 'ล',
    'y' : 'ย',
    'v' : 'ว',
    's' : 'ส',
    'h' : 'ห',
}
v2t = {
    'a': 'ะ',
    'ā': 'า',
    'i': 'ิ',
    'ī': 'ี',
    'u': 'ุ',
    'ū': 'ู',
    #
    'e': 'เ',
    'o': 'โ',
}
````

````
def rom2th(w):
  """ แปลงแบบที่ช่วยให้อ่านง่าย """
  w = w.lower()
  # เติมจุดล่าง กรณี แบบ dd
  w = re.sub('([kgcjtdpbñnmls])\\1', '\\1ฺ\\1', w)
  # กรณี nd
  w = re.sub('([ñnm])([kgcjtdpb])', '\\1ฺ\\2', w)
  w = re.sub('[kgcjtdpb]h?|[ñnmrlyvsh]', (lambda m: c2t[m.group()]), w)
  # ขึ้นต้นด้วยสระ เติม อ ให้
  if w[0] in v2t.keys():
    w = 'อ'+w
  w = re.sub('[aāiīuūoe]', (lambda m: v2t[m.group()]), w)
  w = re.sub('(.)(โ|เ)', '\\2\\1', w)
  # ลบ ะ ทิ้งกรณีมีจุดล่างแล้ว
  w = re.sub('ะ(.ฺ)', 'ั\\1', w)
  # ง มาแทน น
  w = re.sub('น(ฺ[กขคฆ])', 'ง\\1', w)
  # w = w.replace('ะ','')
  return w
````

### Roman -> ไทย

````
def roman2thai(text):
  # แทนที่ combi ก่อน
  for ro, th in sorted(conso_combi, key=lambda x: -len(x[0])):
    text = text.replace(ro, th)
  # กรณีพิเศษที่หลงเหลือ
  for th, ro in sorted(ro_specials, key=lambda x: -len(x[1])):
    text = text.replace(ro, th)
  return text
````

### ไทย -> Roman

````
def thai2roman(text):
  # แทนที่ combi ก่อน
  for ro, th in sorted(conso_combi, key=lambda x: -len(x[1])):
    text = text.replace(th, ro)
  # กรณีพิเศษที่หลงเหลือ
  for th, ro in sorted(ro_specials, key=lambda x: -len(x[0])):
    text = text.replace(th, ro)
  return text
````
