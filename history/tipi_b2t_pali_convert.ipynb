```
tipi = Path('tipi')
```

```
# natural sort
# conv = lambda text: int(text) if text.isdigit() else text.lower()
# alphanum = lambda key: [conv(c) for c in re.split('([0-9]+)', str(key))]
natsort = lambda s: [int(t) if t.isdigit() else t.lower() for t in re.split('(\d+)', str(s))]
```

```
suttas = sorted(tipi.rglob('[abcde]*.txt'), key=natsort)
len(suttas)
```

```
suttas[:15]
```

```
sut = suttas[0]
sut.name[0]   # a คือ ทีฆนิกาย
```

```
# อยากรู้ว่า แต่ละตัว a, b, c อันไหนมีไฟล์เยอะสุด
cnt = Counter(t.name[0] for t in tipi.glob('*.txt'))
cnt
```
